```python
# COMPLETE WORKFLOW

# Import necessary packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read in data
aqi_17 = pd.read_csv('data/daily_aqi_by_county_2017.csv')
aqi_18 = pd.read_csv('data/daily_aqi_by_county_2018.csv')

# Concatenating the two data frames together into a new dataframe
aqi = pd.concat([aqi_17, aqi_18])

# Simplify column names
aqi.columns = (aqi.columns
                  .str.lower()
                  .str.replace(' ','_')
                )

# Making a new data frame that only keeps rows where the county name is Santa Barbara
aqi_sb = aqi[aqi['county_name'] == 'Santa Barbara']

# Adjusting the new data frame so that it only contains certain columns
aqi_sb = aqi_sb[['date', 'aqi', 'category', 'defining_parameter', 'defining_site', 'number_of_sites_reporting']]

# Updating the date column to be a `datetime64` data type instead of an object
aqi_sb['date'] = aqi_sb['date'].astype('datetime64[ns]')

# Updating the data frame so that the date column becomes the index
aqi_sb = aqi_sb.set_index('date')

# Calculate AQI rolling average over 5 days
rolling_average = aqi_sb['aqi'].rolling(window = '5D').mean()

# Adding the rolling average as a new column in the sb data frame
aqi_sb['five_day_average'] = rolling_average


# Making a line plot with the aqi and rolling average
# One line for each
aqi_sb['aqi'].plot(figsize=(10, 6), color = 'steelblue', alpha = 0.5, linewidth = 1, label = 'Daily AQI')
aqi_sb['five_day_average'].plot(color = 'darkorange', linewidth = 2, label = '5-Day Rolling Average')

# Adding a horizontal line at AQI = 100
plt.axhline(100, color = 'black', linestyle = '--', linewidth = 1, label = 'Unhealthy for Sensitive Groups (100)')

# Adding an annotation where the Thomas Fire occurred
plt.annotate('Thomas Fire\n(Dec 2017)', # Label and a new line so it reads better
             xy = (pd.Timestamp('2017-12-10'), 300), # Where the arrow points to. x value is date of Thomas fire and y value is 300
             xytext = (pd.Timestamp('2018-03-01'), 280),  # Where the text sits. x value is timestamp and y value is 280.
             arrowprops = dict(arrowstyle = '->', color = 'gray'))  # Making the arrow

# Adding labels and legends
plt.xlabel('Date')
plt.ylabel('Air Quality Index (AQI)')
plt.title('Daily AQI and 5-Day Rolling Average in Santa Barbara County, 2017–2018')
plt.legend(loc = 'upper left')  # Changing where the legend is on the map
```

In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()` method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as output

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values. 

```python
# Calculate AQI rolling average over 5 days
rolling_average = aqi_sb['aqi'].rolling(window = '5D').mean()
rolling_average
```
